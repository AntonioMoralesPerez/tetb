### Construct the matrix of EBRs in a easy way using the functions we implemented.

You can build the EBR matrix, represented by `A` in the [article](https://arxiv.org/abs/2305.18257), by placing the "symmetry vectors" of EBRs as columns.

We implemented a function `ir2vec` that transform a sum of irreps into a "symmetry vector", so the matrix `A` can be built as

```python
A = np.stack(
    (
        ir2vec(G1p + G2m + R1p + R2m + M1 + X1, irreps),
        ir2vec(G1m + G2p + R1m + R2p + M2 + X2, irreps),
        ir2vec(G3p + G3m + R3p + R3m + M1 + M2 + X1 + X2, irreps),
        ir2vec(G4p + G5m + R4p + R5m + M2 + M3 + M4 + X2 + X3 + X4, irreps),
        ir2vec(G4m + G5p + R4m + R5p + M1 + M3 + M4 + X1 + X3 + X4, irreps),
        ir2vec(G1p + G5p + R1p + R5p + M1 + M3 + X1 + X4, irreps),
        ir2vec(G1m + G5m + R1m + R5m + M2 + M3 + X2 + X4, irreps),
        ir2vec(G2p + G4p + R2p + R4p + M2 + M4 + X2 + X3, irreps),
        ir2vec(G2m + G4m + R2m + R4m + M1 + M4 + X1 + X3, irreps),
        ir2vec(
            G3p + G4p + G5p + R3p + R4p + R5p + M1 + M2 + M3 + M4 + X1 + X2 + X3 + X4,
            irreps,
        ),
        ir2vec(
            G3m + G4m + G5m + R3m + R4m + R5m + M1 + M2 + M3 + M4 + X1 + X2 + X3 + X4,
            irreps,
        ),
        ir2vec(G1p + G5p + R2m + R4m + M1 + M3 + X1 + X3, irreps),
        ir2vec(G1m + G5m + R2p + R4p + M2 + M3 + X2 + X3, irreps),
        ir2vec(G2p + G4p + R1m + R5m + M2 + M4 + X2 + X4, irreps),
        ir2vec(G2m + G4m + R1p + R5p + M1 + M4 + X1 + X4, irreps),
        ir2vec(
            G3p + G4p + G5p + R3m + R4m + R5m + M1 + M2 + M3 + M4 + X1 + X2 + X3 + X4,
            irreps,
        ),
        ir2vec(
            G3m + G4m + G5m + R3p + R4p + R5p + M1 + M2 + M3 + M4 + X1 + X2 + X3 + X4,
            irreps,
        ),
        ir2vec(G1p + G2m + G3p + G3m + R4m + R5p + M1 + M3 + M4 + 2 * X1 + X2, irreps),
        ir2vec(G4p + G5m + R1m + R2p + R3p + R3m + M1 + 2 * M2 + X2 + X3 + X4, irreps),
        ir2vec(G1m + G2p + G3p + G3m + R4p + R5m + M2 + M3 + M4 + X1 + 2 * X2, irreps),
        ir2vec(G4m + G5p + R1p + R2m + R3p + R3m + 2 * M1 + M2 + X1 + X3 + X4, irreps),
        ir2vec(
            G1p
            + G1m
            + G3p
            + G3m
            + G5p
            + G5m
            + R4p
            + R4m
            + R5p
            + R5m
            + M1
            + M2
            + 3 * M3
            + M4
            + 2 * X1
            + 2 * X2
            + X3
            + X4,
            irreps,
        ),
        ir2vec(
            G2p
            + G2m
            + G3p
            + G3m
            + G4p
            + G4m
            + R4p
            + R4m
            + R5p
            + R5m
            + M1
            + M2
            + M3
            + 3 * M4
            + 2 * X1
            + 2 * X2
            + X3
            + X4,
            irreps,
        ),
        ir2vec(
            G4p
            + G4m
            + G5p
            + G5m
            + R1p
            + R1m
            + R3p
            + R3m
            + R5p
            + R5m
            + 2 * M1
            + 2 * M2
            + M3
            + M4
            + X1
            + X2
            + X3
            + 3 * X4,
            irreps,
        ),
        ir2vec(
            G4p
            + G4m
            + G5p
            + G5m
            + R2p
            + R2m
            + R3p
            + R3m
            + R4p
            + R4m
            + 2 * M1
            + 2 * M2
            + M3
            + M4
            + X1
            + X2
            + 3 * X3
            + X4,
            irreps,
        ),
    ),
    axis=1,
)
```

```python
print(A)
```

Thanks to this matrix you can transform linear combinations of EBRs into symmetry vectors or linear combinations of irreps using built-in functions

```python
print(ebr2irvec(v=A2g_4b + Eu_4b, ebrs=ebrs, EBR=A))
```

```python
ebr2ir(A2g_4b + Eu_4b, ebrs=ebrs, EBR=A, irreps=irreps)
```

## **Main problem :** Find a TETB model from a vector of irreps obtained from the bands of the crystal

First you need to define the vector of irreps you are working with. This can be easily done with built-in functions

```python
vec_test = (
    G1p
    + G1m
    + G2p
    + G5p
    + R1m
    + R2p
    + R4m
    + R5p
    + M1
    + M2
    + M3
    + M4
    + X1
    + X2
    + X3
    + X4
)
vec_test = ir2vec(v=vec_test, irreps=irreps)
print(vec_test)
```

With this you can obtain all posible solutions in a list form or only the physical ones. We show here both posibilites for `t=2` number of auxiliary bands

```python
result1 = showAllResults(
    v=vec_test,
    t=2,
    dim=dim_ebrs,
    EBRs=A,
    N_gamma=N_gamma,
    ebrs=ebrs,
    irreps=irreps,
    SG=224,
)
result1
```

```python
result2 = showOnlyPhysical(
    v=vec_test,
    t=2,
    dim=dim_ebrs,
    EBRs=A,
    N_gamma=N_gamma,
    ebrs=ebrs,
    irreps=irreps,
    SG=224,
)
result2
```

Now let us focus on the case in the case in [article](https://arxiv.org/abs/2305.18257)

```python
chiara = G4m + G2m + R4m + R5p + M1 + 2 * M4 + X1 + X3 + X4
chiara = ir2vec(v=chiara, irreps=irreps)
print(chiara)
```
